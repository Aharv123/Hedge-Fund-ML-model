```python
import pandas as pd

# Load quarterly CSVs
balance_sheet_quarterly = pd.read_csv('data/balanceSheetHistory_quarterly.csv')
cashflow_quarterly = pd.read_csv('data/cashflowStatement_quarterly.csv')
income_statement_quarterly = pd.read_csv('data/incomeStatementHistory_quarterly.csv')

# Inspect and clean the data
print(balance_sheet_quarterly.isna().sum())
print(cashflow_quarterly.isna().sum())
print(income_statement_quarterly.isna().sum())

# Fill NaN values - adjust based on context (forward fill in this case)
balance_sheet_quarterly.fillna(method='ffill', inplace=True)
cashflow_quarterly.fillna(method='ffill', inplace=True)
income_statement_quarterly.fillna(method='ffill', inplace=True)

# Merge the datasets based on 'stock' and 'endDate'
merged_quarterly = balance_sheet_quarterly.merge(cashflow_quarterly, on=['stock', 'endDate'], how='inner')
merged_quarterly = merged_quarterly.merge(income_statement_quarterly, on=['stock', 'endDate'], how='inner')

# Ensure no missing values after merging
print(merged_quarterly.isna().sum())

# Save the cleaned and merged dataset
merged_quarterly.to_csv('new_data/new_balance_cashflow_incom_quarterlymerged.csv', index=False)

# Display first few rows to verify
print(merged_quarterly.head())
```
