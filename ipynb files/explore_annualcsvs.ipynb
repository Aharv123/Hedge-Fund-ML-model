```python
import pandas as pd

# Load annual CSVs
balance_sheet_annual = pd.read_csv('data/balanceSheetHistory_annually.csv')
cashflow_annual = pd.read_csv('data/cashflowStatement_annually.csv')
income_statement_annual = pd.read_csv('data/incomeStatementHistory_annually.csv')

# Inspect and clean the data
print(balance_sheet_annual.isna().sum())
print(cashflow_annual.isna().sum())
print(income_statement_annual.isna().sum())

# Fill NaN values - you can fill with a method that makes sense for the column, like forward fill, zero, or the mean
balance_sheet_annual.fillna(method='ffill', inplace=True)
cashflow_annual.fillna(method='ffill', inplace=True)
income_statement_annual.fillna(method='ffill', inplace=True)

# Merge the datasets based on 'stock' and 'endDate'
merged_annual = balance_sheet_annual.merge(cashflow_annual, on=['stock', 'endDate'], how='inner')
merged_annual = merged_annual.merge(income_statement_annual, on=['stock', 'endDate'], how='inner')

# Ensure no missing values after merging
print(merged_annual.isna().sum())

# Save the cleaned and merged dataset
merged_annual.to_csv('new_data/new_balance_cashflow_incom_annualmerged.csv', index=False)

# Display first few rows to verify
print(merged_annual.head())
```
